```python
# Dictionary Collection
tickers = {'aapl':'Apple Inc', 'FB':'Facebook', 'GOOG':'Alphabet'} # curly braces make this a dictionary
type(tickers)
tickers['FB'] # we can acess values by their key
tickers['IBM'] = 'Herman Hollerith'
tickers['IBM'] = 'International Business Machines'
tickers
```

```python
# loops
for i in myStocks:
    print( i, type(i) )

for i in prices:
    print(i)
```

```python
r = range(-10,10,3) # start, stop-before, step=1
for i in r:
    print(i**2) # we have + - * / also ** raise to power
```

```python
# we may iterate over a dict
for (key, value) in tickers.items():
    print(key, value)
```

```python
# Conditional logic
# > < == != and or <= >=
a = 3
b = 9
if a > b:
    print('greater')
elif a < b: # else if...
    print('less')
else:
    print('unknown')
```

```python
# functions
def calcDivi(n, p):
    '''We can write a docstring about our function
    Here we calculate a percentage dividend based on n and p'''
    dividend = n*p
    return dividend

d1 = calcDivi(100, 0.05)
print(d1)
    
```
