## Numpy and Pandas

```python
# we need both mupy and pandas
import numpy as np
import pandas as pd
```

```python
# Numpy includes an array data type
a = np.array([1,2,3,4,5])
a*8
b = np.array([8,7,6,5,4])
b*a
```

```python
# Stats
# np.random.seed(42)
c = np.random.normal(5,2) # mean of 5 sd of 2
c
```

```python
costs = np.random.normal(5, 2, 100) ## mean=5 sd=2 100 items
costs
```

```python
# stats
costs.mean()
print(costs.max())
costs.min()
costs.std()
costs.sum()
## only the very last operation is shown in jupyter output
```

```python
# formatting printed output
print( f'Mean: {costs.mean():0.3f} min:{costs.min()}'  )
```

```python
# reading in external data
# NB this may be a local file or a file on a network
# sp500 = pd.read_csv('StockData/SP500.csv')
sp500= pd.read_csv('https://raw.githubusercontent.com/onionmccabbage/pythonOneSept2026/refs/heads/main/StockData/SP500.csv')
sp500
```

```python
# sp500.info()
# we often need to convert a string into a date
sp500['Date'] = pd.to_datetime( sp500['Date'], format='%m/%d/%y' ) # we specify the current format
sp500
```

```python
# We can use any column to be an index
sp500.set_index(['Date'], inplace=True)
```
